```python
import heapq

def greedy_best_first_search(graph, heuristic, start, goal):
    """
    Implements the Greedy Best-First Search algorithm.

    Args:
        graph (dict): A dictionary representing the graph. Keys are nodes,
                      values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 1}}
        heuristic (dict): A dictionary of heuristic values for each node
                          estimating the cost to the goal.
                          Example: {'A': 10, 'B': 7}
        start: The starting node.
        goal: The goal node.

    Returns:
        tuple: A tuple containing the path from start to goal (list of nodes)
               and the total estimated cost, or (None, None) if no path is found.
    """
    # Priority queue stores (heuristic_cost, current_node, path)
    # The heuristic_cost is used for prioritization in greedy search.
    priority_queue = [(heuristic[start], start, [start])]
    visited = set()

    while priority_queue:
        h_cost, current_node, path = heapq.heappop(priority_queue)

        if current_node == goal:
            # In greedy best-first search, we return the path and the heuristic
            # cost of the goal node (which should be 0 if the heuristic is consistent).
            return path, h_cost

        if current_node in visited:
            continue

        visited.add(current_node)

        for neighbor, cost in graph.get(current_node, {}).items():
            if neighbor not in visited:
                new_path = path + [neighbor]
                # In greedy best-first search, we only care about the heuristic
                # of the neighbor for prioritization, not the path cost so far.
                heapq.heappush(priority_queue, (heuristic[neighbor], neighbor, new_path))

    return None, None # No path found

# Define a simple graph
# Note: Edge costs are included for completeness but not directly used for
# prioritization in basic greedy best-first search (only heuristic is used).
# They would be used in A* search.
graph = {
    'A': {'B': 1, 'C': 1},
    'B': {'D': 1, 'E': 1},
    'C': {'F': 1},
    'D': {},
    'E': {'G': 1},
    'F': {'G': 1},
    'G': {}
}

# Define heuristic values (estimated cost from node to goal 'G')
heuristic = {
    'A': 10,
    'B': 7,
    'C': 6,
    'D': 8,
    'E': 9,
    'F': 1,
    'G': 0
}

# Example usage
start_node = 'A'
goal_node = 'G'

path, estimated_cost = greedy_best_first_search(graph, heuristic, start_node, goal_node)

if path:
    print(f"Path from {start_node} to {goal_node}: {path}")
    print(f"Heuristic value at goal: {estimated_cost}") # This should be 0 if heuristic is correct for goal
else:
    print(f"No path found from {start_node} to {goal_node}")

# Another example: If start is C, path should be C -> F -> G
start_node_2 = 'C'
path_2, estimated_cost_2 = greedy_best_first_search(graph, heuristic, start_node_2, goal_node)
if path_2:
    print(f"\nPath from {start_node_2} to {goal_node}: {path_2}")
    print(f"Heuristic value at goal: {estimated_cost_2}")
else:
    print(f"No path found from {start_node_2} to {goal_node}")
```

```python
from collections import deque

def breadth_first_search(graph, start, goal):
    """
    Implements the Breadth-First Search (BFS) algorithm.

    Args:
        graph (dict): A dictionary representing the graph. Keys are nodes,
                      values are dictionaries of neighbors and their costs.
                      Example: {'A': {'B': 1, 'C': 1}}
        start: The starting node.
        goal: The goal node.

    Returns:
        list: A list of nodes representing the path from start to goal,
              or None if no path is found.
    """
    queue = deque([(start, [start])]) # Stores (current_node, path_to_current_node)
    visited = set()

    while queue:
        current_node, path = queue.popleft()

        if current_node == goal:
            return path

        if current_node in visited:
            continue

        visited.add(current_node)

        for neighbor in graph.get(current_node, {}).keys():
            if neighbor not in visited:
                new_path = path + [neighbor]
                queue.append((neighbor, new_path))

    return None # No path found

# Using the same graph defined previously
# graph = {
#     'A': {'B': 1, 'C': 1},
#     'B': {'D': 1, 'E': 1},
#     'C': {'F': 1},
#     'D': {},
#     'E': {'G': 1},
#     'F': {'G': 1},
#     'G': {}
# }

# Example usage for BFS
start_node_bfs = 'A'
goal_node_bfs = 'G'

path_bfs = breadth_first_search(graph, start_node_bfs, goal_node_bfs)

if path_bfs:
    print(f"BFS Path from {start_node_bfs} to {goal_node_bfs}: {path_bfs}")
else:
    print(f"No BFS path found from {start_node_bfs} to {goal_node_bfs}")

start_node_bfs_2 = 'E'
path_bfs_2 = breadth_first_search(graph, start_node_bfs_2, goal_node_bfs)

if path_bfs_2:
    print(f"BFS Path from {start_node_bfs_2} to {goal_node_bfs}: {path_bfs_2}")
else:
    print(f"No BFS path found from {start_node_bfs_2} to {goal_node_bfs}")
```

```python
def depth_first_search(graph, start, goal):
    """
    Implements the Depth-First Search (DFS) algorithm.

    Args:
        graph (dict): A dictionary representing the graph. Keys are nodes,
                      values are dictionaries of neighbors and their costs.
        start: The starting node.
        goal: The goal node.

    Returns:
        list: A list of nodes representing the path from start to goal,
              or None if no path is found.
    """
    stack = [(start, [start])] # Stores (current_node, path_to_current_node)
    visited = set()

    while stack:
        current_node, path = stack.pop() # Use pop() for LIFO (stack behavior)

        if current_node == goal:
            return path

        if current_node in visited:
            continue

        visited.add(current_node)

        # Iterate through neighbors in reverse order if you want to match the output
        # of a typical left-to-right DFS exploration when appending to a stack
        for neighbor in sorted(graph.get(current_node, {}).keys(), reverse=True):
            if neighbor not in visited:
                new_path = path + [neighbor]
                stack.append((neighbor, new_path))

    return None # No path found

# Using the same graph defined previously
# graph = {
#     'A': {'B': 1, 'C': 1},
#     'B': {'D': 1, 'E': 1},
#     'C': {'F': 1},
#     'D': {},
#     'E': {'G': 1},
#     'F': {'G': 1},
#     'G': {}
# }

# Example usage for DFS
start_node_dfs = 'A'
goal_node_dfs = 'G'

path_dfs_result = depth_first_search(graph, start_node_dfs, goal_node_dfs)

if path_dfs_result:
    print(f"DFS Path from {start_node_dfs} to {goal_node_dfs}: {path_dfs_result}")
else:
    print(f"No DFS path found from {start_node_dfs} to {goal_node_dfs}")

start_node_dfs_2 = 'E'
path_dfs_result_2 = depth_first_search(graph, start_node_dfs_2, goal_node_dfs)

if path_dfs_result_2:
    print(f"DFS Path from {start_node_dfs_2} to {goal_node_dfs}: {path_dfs_result_2}")
else:
    print(f"No DFS path found from {start_node_dfs_2} to {goal_node_dfs}")
```
