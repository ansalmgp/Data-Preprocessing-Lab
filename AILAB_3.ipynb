```python
import heapq

def a_star(graph, heuristic, start, goal):
    pq = [(heuristic[start], 0, start, [start])]  # (f, g, node, path)
    visited = set()

    while pq:
        f, g, current, path = heapq.heappop(pq)

        if current == goal:
            return path, g

        if current in visited:
            continue

        visited.add(current)

        for neighbor, cost in graph[current]:
            if neighbor not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(pq, (new_f, new_g, neighbor, path + [neighbor]))

    return None, float("inf")


# Example Graph
graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('A', 1), ('D', 3), ('E', 6)],
    'C': [('A', 3), ('F', 5)],
    'D': [('B', 3)],
    'E': [('B', 6), ('F', 2)],
    'F': [('C', 5), ('E', 2)]
}

# Heuristic values (estimated cost to goal F)
heuristic = {
    'A': 6,
    'B': 5,
    'C': 4,
    'D': 6,
    'E': 2,
    'F': 0
}

start = 'A'
goal = 'F'

path, cost = a_star(graph, heuristic, start, goal)

print("Shortest Path:", " -> ".join(path))
print("Total Cost:", cost)
```

```python
import heapq

def uniform_cost_search(graph, start, goal):
    # Priority Queue: (cost, current_node, path)
    priority_queue = [(0, start, [start])]
    visited = set()

    while priority_queue:
        cost, current, path = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        # Goal reached
        if current == goal:
            return path, cost

        # Explore neighbors
        for neighbor, edge_cost in graph[current]:
            if neighbor not in visited:
                heapq.heappush(
                    priority_queue,
                    (cost + edge_cost, neighbor, path + [neighbor])
                )

    return None, float('inf')


# Graph Representation
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('C', 2), ('D', 5)],
    'C': [('A', 4), ('B', 2), ('D', 1)],
    'D': [('B', 5), ('C', 1)]
}

# Input
start = 'A'
goal = 'D'

# Run UCS
path, cost = uniform_cost_search(graph, start, goal)

# Output
if path:
    print("Shortest Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found.")
```
