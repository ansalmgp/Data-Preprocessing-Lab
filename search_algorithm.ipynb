1. Uniform Cost Search (UCS) — Uninformed Search


```python
import heapq

def uniform_cost_search(graph, start, goal):

    priority_queue = [(0, start, [])]
    visited = set()

    while priority_queue:

        cost, node, path = heapq.heappop(priority_queue)

        if node in visited:
            continue

        path = path + [node]
        visited.add(node)

        # Goal reached
        if node == goal:
            return cost, path

        for neighbor, edge_cost in graph[node]:
            if neighbor not in visited:
                heapq.heappush(
                    priority_queue,
                    (cost + edge_cost, neighbor, path)
                )

    return None


graph = {
    'A': [('B', 2), ('C', 4)],
    'B': [('D', 3)],
    'C': [('D', 1)],
    'D': []
}


result = uniform_cost_search(graph, 'A', 'D')

print("Minimum Cost:", result[0])
print("Path:", result[1])
```

A* Search Algorithm (Informed Search)


```python
import heapq


def a_star(graph, heuristic, start, goal):

    queue = [(0, start, [])]

    cost = {start: 0}

    while queue:

        f, node, path = heapq.heappop(queue)

        path = path + [node]

        if node == goal:
            return cost[node], path


        for neighbor, edge_cost in graph[node]:

            new_cost = cost[node] + edge_cost


            if neighbor not in cost or new_cost < cost[neighbor]:

                cost[neighbor] = new_cost

                total_cost = (
                    new_cost + heuristic[neighbor]
                )

                heapq.heappush(
                    queue,
                    (total_cost, neighbor, path)
                )

    return None



graph = {
    'A': [('B',2), ('C',4)],
    'B': [('D',3)],
    'C': [('D',1)],
    'D': []
}


heuristic = {
    'A':5,
    'B':3,
    'C':1,
    'D':0
}


result = a_star(graph, heuristic, 'A', 'D')

print("Minimum Cost:", result[0])
print("Path:", result[1])
```
